```python
df_clean["Amount_Category"] = pd.cut(
    df_clean["Amount"],
    bins=[-1, 50, 200, 500, float("inf")],
    labels=["Low", "Medium", "High", "Very High"]
)
```

```python
amount_category_analysis = df_clean.groupby(
    "Amount_Category",
    observed=False
)["Class"].agg(
    ["sum", "count", "mean"]
)

amount_category_analysis["Fraud_Rate"] = (
    amount_category_analysis["mean"] * 100
)

print(amount_category_analysis)
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    amount_category_analysis.index.astype(str),
    amount_category_analysis["Fraud_Rate"]
)

plt.title("Fraud Rate by Transaction Amount Category")
plt.xlabel("Transaction Amount Category")
plt.ylabel("Fraud Rate (%)")
plt.show()
```

```python
# Select only numerical columns
numeric_df = df_clean.select_dtypes(include="number")

# Calculate correlations
correlation = numeric_df.corr()

# Get correlations with the fraud label
fraud_correlation = correlation["Class"].sort_values(
    ascending=False
)

print(fraud_correlation)
```

```python
# Remove Class itself from the correlation results
feature_corr = fraud_correlation.drop("Class")

# Select 5 strongest positive and 5 strongest negative features
top_features = pd.concat([
    feature_corr.nlargest(5),
    feature_corr.nsmallest(5)
]).sort_values()

# Create chart
plt.figure(figsize=(10, 6))

plt.barh(
    top_features.index,
    top_features.values
)

plt.axvline(0, linewidth=1)
plt.title("Top Features Correlated with Credit Card Fraud")
plt.xlabel("Correlation with Fraud (Class)")
plt.ylabel("Features")
plt.grid(axis="x")
plt.show()
```

```python
df_clean["Risk_Score"] = (
    (df_clean["Amount"] > 500).astype(int) +
    (df_clean["V14"] < -5).astype(int) +
    (df_clean["V17"] < -4).astype(int) +
    (df_clean["V11"] > 2).astype(int)
)
```

```python
def assign_alert(score):
    if score >= 3:
        return "High"
    elif score == 2:
        return "Medium"
    elif score == 1:
        return "Low"
    else:
        return "Normal"

df_clean["Alert_Level"] = df_clean["Risk_Score"].apply(assign_alert)
```

```python
alert_distribution = df_clean["Alert_Level"].value_counts()
print(alert_distribution)
```

```python
alert_performance = pd.crosstab(
    df_clean["Alert_Level"],
    df_clean["Class"]
)

print(alert_performance)
```

```python
alert_summary = df_clean.groupby("Alert_Level").agg(
    Total_Transactions=("Class", "count"),
    Fraud_Transactions=("Class", "sum")
)

alert_summary["Fraud_Rate"] = (
    alert_summary["Fraud_Transactions"] /
    alert_summary["Total_Transactions"] * 100
)

print(alert_summary)
```

```python
plt.figure(figsize=(9, 5))

plt.bar(
    alert_summary.index,
    alert_summary["Fraud_Rate"]
)

plt.title("Fraud Rate by Alert Level")
plt.xlabel("Alert Level")
plt.ylabel("Fraud Rate (%)")
plt.show()
```

```python
alert_order = ["Normal", "Low", "Medium", "High"]

alert_summary = alert_summary.reindex(alert_order)

plt.figure(figsize=(9, 5))
plt.bar(alert_summary.index, alert_summary["Fraud_Rate"])
plt.title("Fraud Rate by Alert Level")
plt.xlabel("Alert Level")
plt.ylabel("Fraud Rate (%)")
plt.show()
```
